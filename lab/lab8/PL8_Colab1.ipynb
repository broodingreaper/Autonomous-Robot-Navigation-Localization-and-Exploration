### What is Machine Learning (ML)?
Machine learning figures things out on its own. It doesn't need explicit programming to find patterns, features, or irregularities in big data sets. Compare the traditional programming paradigm to the machine learning paradigm below.

For a Machine Learning algorithm to solve a problem, there are two key features. (1) it will take a piece of **data** and (2) will take a **guess** at the **answer**. Then, it will *optimize* its **guess** until it gets the correct **answer**. The concept of *loss* will be what helps to *optimize* the **guess**.

### Task 1.4

Looking at your visual graphs, what evidence do you have to support your belief whether the model is fit/underfit/overfit? Why does the model exhibit those fit-ness characteristics?



---

Write down your answer here for Task 1.4
The evidence for whether a model is underfit, overfit, or well-fit comes from how the decision boundary (colored regions) compares to the data points. In the underfit case, the boundary is overly simple (often almost straight) and fails to separate the orange and purple points correctly, showing many misclassified regions. This happens because the model has too few neurons or layers, so it lacks the capacity to learn complex patterns. In the overfit case, the boundary becomes very irregular and tightly wraps around individual data points, even small noise or outliers, which indicates the model is memorizing the data rather than generalizing. This occurs when there are too many neurons, layers, or excessive training. In the well-fit case, the boundary is smooth and separates the two classes accurately while ignoring small noise, showing that the model has the right balance of complexity to generalize from the data.
